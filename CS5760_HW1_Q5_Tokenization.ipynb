# Homework 1 – Q5 Tokenization

This notebook demonstrates naive tokenization, manual tokenization, and tool-based (regex) tokenization with visible outputs.

## Paragraph

```python
paragraph = """Natural Language Processing is very interesting.
It helps computers understand human language.
Tokenization is the first step in NLP."""
print(paragraph)
```

## 1. Naïve Space-Based Tokenization

```python
naive_tokens = paragraph.split()
naive_tokens
```

## 2. Manual Tokenization (Punctuation-Aware)

```python
import re
manual_tokens = re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?|[.,!?]", paragraph)
manual_tokens
```

## 3. Comparison: Naïve vs Manual

```python
print('Naive Tokens:')
print(naive_tokens)

print('\nManual Tokens:')
print(manual_tokens)
```

## 4. Tool-Based Tokenization (Regex Tokenizer)

```python
tool_tokens = re.findall(r"\w+|[^\w\s]", paragraph)
tool_tokens
```

## 5. Comparison: Manual vs Tool

```python
print('Manual Tokens:')
print(manual_tokens)

print('Tool Tokens:')
print(tool_tokens)
```

## 6. Multiword Expressions (MWEs)

```python
MWEs = [
    'Natural Language Processing',
    'human language',
    'first step'
]
MWEs
```

## 7. Reflection
The hardest part of tokenization is handling punctuation and multiword expressions.
English tokenization is easier than morphologically rich languages.
Punctuation creates ambiguity in token boundaries.
MWEs should be treated as single units.
Manual tokenization is accurate but time-consuming.
Rule-based tools are fast but less flexible.
